```python
import pandas as pd
data = {
    'Name': ['Alice', 'Bob', 'Carol', 'Dave', 'Eva', 'Frank', 'Grace'],
    'Age': [23, 25, 22, 24, 21, 26, 23],
    'Gender': ['F', 'M', 'F', 'M', 'F', 'M', 'F'],
    'Marks': [85, 58, 90, 45, 76, 67, 92],
    'Department': ['CSE', 'ECE', 'CSE', 'ME', 'ECE', 'CSE', 'CSE']
}
df = pd.DataFrame(data)
print("Shape:", df.shape)
print("\nInfo:")
print(df.info())
print("\nSummary Statistics:")
print(df.describe())
```

```python
df['Result'] = df['Marks'].apply(lambda x: 'Pass' if x >= 60 else 'Fail')
print(df[['Result']])
```

```python
def grade(marks):
    if marks >= 90:
        return 'A'
    elif marks >= 80:
        return 'B'
    elif marks >= 70:
        return 'C'
    elif marks >= 60:
        return 'D'
    else:
        return 'F'

df['Grade'] = df['Marks'].apply(grade)
print(df[['Grade']])
```

```python
grouped = df.groupby('Department')
print(grouped['Marks'].mean())
print(grouped['Name'].count())
```

```python
female_cse = df[(df['Gender'] == 'F') & (df['Department'] == 'CSE') & (df['Marks'] > 80)]
print(female_cse)
```
