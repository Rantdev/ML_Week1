```python
import pandas as pd
import matplotlib.pyplot as plt
data = {
    'Name': ['Alice', 'Bob', 'Carol', 'Dave', 'Eva', 'Frank', 'Grace'],
    'Age': [23, 25, 22, 24, 21, 26, 23],
    'Gender': ['F', 'M', 'F', 'M', 'F', 'M', 'F'],
    'Marks': [85, 58, 90, 45, 76, 67, 92],
    'Department': ['CSE', 'ECE', 'CSE', 'ME', 'ECE', 'CSE', 'CSE']
}
df = pd.DataFrame(data)
df['Result'] = df['Marks'].apply(lambda x: 'Pass' if x >= 60 else 'Fail')
def grade(m):
    if m >= 90:
        return 'A'
    elif m >= 80:
        return 'B'
    elif m >= 70:
        return 'C'
    elif m >= 60:
        return 'D'
    else:
        return 'F'

df['Grade'] = df['Marks'].apply(grade)
plt.figure(figsize=(6, 4))
df['Department'].value_counts().plot(kind='bar')
plt.title('Number of Students per Department')
plt.xlabel('Department')
plt.ylabel('Number of Students')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(5, 5))
df['Result'].value_counts().plot(kind='pie',autopct='%1.1f%%')
plt.title('Result Distribution (Pass vs Fail)')
plt.ylabel('')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(6, 4))
plt.hist(df['Marks'], bins=10)
plt.title('Histogram of Marks')
plt.xlabel('Marks')
plt.ylabel('Number of Students')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 4))
plt.plot(df['Name'], df['Marks'], marker='o', linestyle='-')
plt.title('Student Marks')
plt.xlabel('Student Name')
plt.ylabel('Marks')
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
fig, axs = plt.subplots(1,3,  figsize=(18, 4))
axs[0].hist(df['Marks'], bins=10, )
axs[0].set_title('Histogram of Marks')
axs[0].set_xlabel('Marks')
axs[0].set_ylabel('Count')
grade_counts = df['Grade'].value_counts()
axs[1].pie(grade_counts, labels=grade_counts.index, autopct='%1.1f%%')
axs[1].set_title('Grade Distribution')
dept_counts = df['Department'].value_counts()
axs[2].bar(dept_counts.index, dept_counts.values)
axs[2].set_title('Students per Department')
axs[2].set_xlabel('Department')
axs[2].set_ylabel('Count')

plt.tight_layout()
plt.show()
```
