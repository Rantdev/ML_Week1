```python
students = ((101, "Alice", 20),(102, "Bob", 21),(103, "Charlie", 19))
print("Student Information:")
print("{:<12} {:<15} {:<5}".format("Roll Number", "Name", "Age"))

for student in students:
    roll_number, name, age = student
    print("{:<12} {:<15} {:<5}".format(roll_number, name, age))
```
